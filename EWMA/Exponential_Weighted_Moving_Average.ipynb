```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df =pd.read_csv('DailyDelhiClimateTest.csv')
df.head()
```

```python
plt.scatter(df['date'],df['meantemp'],color ='red')
```

```python
X1 =df['meantemp'].ewm(alpha =0.9).mean() # aplha = 1-beta this is formula  thats why beta is 1 - alpha
```

```python
X1
```

```python
df['ewma'] =X1
df
```

```python
plt.scatter(df['date'],df['meantemp'],color ='red')
plt.plot(df['date'],X1,color='black',label='alpha =0.9')
plt.legend()
plt.show()
```

```python
plt.scatter(df['date'],df['humidity'],color ='green')
plt.show()
```

```python
X2 =df['humidity'].ewm(alpha=0.9).mean()
```

```python
X2
```

```python
df['humidity_ewm'] =X2
df
```

```python
plt.scatter(df['date'],df['humidity_ewm'],color ='red')
plt.plot(df['date'],X1,color='black',label='alpha =0.9')
plt.legend()
plt.show()
```
