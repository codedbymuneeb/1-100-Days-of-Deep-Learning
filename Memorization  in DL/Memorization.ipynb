Fibonacci Series ek aisi number series hoti hai jisme har number, pichlay do numbers ka sum hota hai

```python
import time
def fib(n):
  # Corrected base cases for the Fibonacci sequence
  if n == 0:
    return 0
  elif n == 1:
    return 1
  else:
    return fib(n-1) + fib(n-2)
```

```python
fib(12)
```

```python
start =time.time()
fib(40)
print(time.time()-start)
```

```python
start = time.time()
fib(10)
print(time.time()-start)
```

```python
def fibb(n,d):
  if n in d:
    return d[n]
  else:
    d[n] =fibb(n-1,d) +fibb(n-2,d)
    return d[n]
```

```python
start =time.time()
d ={0:1,1:1}
fibb(40,d)
print(time.time()-start)
```
