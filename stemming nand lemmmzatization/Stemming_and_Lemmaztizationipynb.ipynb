```python

words = ["playing", "played", "plays", "studies", "better", "running"]
```

```python
stemmer = PorterStemmer()
```

```python
for word in words:
  print(f'stemming word {stemmer.stem(word)}')
```

```python
word1 =['university','runnning','going','joking','fishing']
```

```python
for word in word1:
  print(f'stemming word {stemmer.stem(word)}')
```

```python
lemmaztizer =WordNetLemmatizer()
```

```python
for word in words:
  print(f'lemmaztizer word {lemmaztizer.lemmatize(word, pos="v")}')
```
